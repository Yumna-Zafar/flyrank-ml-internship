## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Question: Given a limited reviewer capacity, which content pages should be reviewed
first for a refresh?

Decision improved: which pages a human content reviewer spends their limited time on
this cycle.

Action taken: a reviewer opens the top-ranked pages and decides whether to refresh
(update content), leave alone, or escalate for a bigger rewrite.

Cost of a wrong call:
- False positive (flagged as worth reviewing, but it wasn't): wastes a reviewer's
  limited time on a page that didn't need attention -- an opportunity cost, since that
  time could have gone to a page that actually needed help.
- False negative (a genuinely declining page never gets flagged): the page keeps
  losing visibility/traffic unnoticed, which is the more expensive failure since it
  compounds silently until someone happens to check it manually.

Because false negatives are more costly here, a strong evaluation should weigh recall
alongside precision, not just Precision@K alone.

```python
print("Decision: which pages get reviewed first")
print("Action: refresh / leave alone / escalate")
print("Costlier error: false negative (missed decline)")
```
